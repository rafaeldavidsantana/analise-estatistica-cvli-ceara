```python
import pandas as pd
df = pd.read_excel("CVLI.xlsx")
print(df.columns)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# Lista com as variáveis solicitadas
variaveis_freq = ['Natureza', 'Meio Empregado', 'Gênero', 'Escolaridade da Vítima', 'Raça da Vítima', 'Dia da Semana']

print("--- 1. Tabelas de Frequências ---")
for var in variaveis_freq:
    print(f"\n➔ Tabela para: {var}")
    # Frequência Simples
    freq_simples = df[var].value_counts()
    # Frequência Relativa (%)
    freq_relativa = df[var].value_counts(normalize=True) * 100

    # Criando o DataFrame para exibir bonito
    tabela = pd.DataFrame({
        'Frequência Simples': freq_simples,
        'Frequência Relativa (%)': freq_relativa
    })
    # Aplica a formatação de string com 2 casas decimais e adiciona o símbolo %
    display(tabela.style.format({
        'Frequência Relativa (%)': "{:.2f}%"
    }))
```

```python
print("--- 2. Tabelas Cruzadas ---")

# Meio Empregado vs Gênero
print("\n➔ Meio Empregado vs Gênero (Frequência Simples):")
display(pd.crosstab(df['Meio Empregado'], df['Gênero']))

print("\n➔ Meio Empregado vs Gênero (Frequência Relativa %):")
freq_rel_genero = pd.crosstab(df['Meio Empregado'], df['Gênero'], normalize='all') * 100
# Aplica a formatação de string com 2 casas decimais e adiciona o símbolo %
display(freq_rel_genero.style.format("{:.2f}%"))

# Escolaridade vs Raça
print("\n➔ Escolaridade da Vítima vs Raça da Vítima (Frequência Simples):")
display(pd.crosstab(df['Escolaridade da Vítima'], df['Raça da Vítima']))

print("\n➔ Escolaridade da Vítima vs Raça da Vítima (Frequência Relativa %):")
freq_rel_raca = pd.crosstab(df['Escolaridade da Vítima'], df['Raça da Vítima'], normalize='all') * 100
# Aplica a formatação de string com 2 casas decimais e adiciona o símbolo %
display(freq_rel_raca.style.format("{:.2f}%"))
```

```python
import seaborn as sns
# Configurando o tamanho da figura
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# --- Gráficos de Setor (Pizza) ---
# Gênero
df['Gênero'].value_counts().plot.pie(ax=axes[0, 0], autopct='%1.1f%%', startangle=90, cmap='Pastel1')
axes[0, 0].set_title('Gráfico de Setor - Gênero')
axes[0, 0].set_ylabel('')

# Meio Empregado (Limitando aos top 5 para o gráfico não ficar confuso)
df['Meio Empregado'].value_counts().head(5).plot.pie(ax=axes[0, 1], autopct='%1.1f%%', startangle=90, cmap='Set3')
axes[0, 1].set_title('Gráfico de Setor - Meio Empregado (Top 5)')
axes[0, 1].set_ylabel('')

# --- Gráficos de Barra ---
# Dia da Semana
sns.countplot(data=df, x='Dia da Semana', ax=axes[1, 0], palette='Blues_r', order=df['Dia da Semana'].value_counts().index)
axes[1, 0].set_title('Gráfico de Barra - Dia da Semana')
axes[1, 0].tick_params(axis='x', rotation=45)

# Escolaridade
sns.countplot(data=df, x='Escolaridade da Vítima', ax=axes[1, 1], palette='Oranges_r', order=df['Escolaridade da Vítima'].value_counts().index)
axes[1, 1].set_title('Gráfico de Barra - Escolaridade da Vítima')
axes[1, 1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
# 1. Convertendo a coluna 'Data'
df['Data_Convertida'] = pd.to_datetime(df['Data'], dayfirst=True, errors='coerce')
df['Mês'] = df['Data_Convertida'].dt.month
df['Ano'] = df['Data_Convertida'].dt.year

# 2. Arrumando a coluna 'Hora'
# Transforma em texto, separa no primeiro ':' e junta com ':00' (Ex: '21:45:00' vira '21:00')
df['Hora_Agrupada'] = df['Hora'].astype(str).str.split(':').str[0] + ':00'

# Limpa valores em branco/nulos que podem ter virado 'nan:00'
df_limpo = df[df['Hora_Agrupada'] != 'nan:00']

# 3. Criando os gráficos
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Gráfico Linha - Mês
df['Mês'].value_counts().sort_index().plot(kind='line', marker='o', ax=axes[0], color='blue')
axes[0].set_title('Ocorrências por Mês')
axes[0].set_xlabel('Mês')
axes[0].set_xticks(range(1, 13)) # Garante que mostre os meses de 1 a 12

# Gráfico Linha - Ano
df['Ano'].value_counts().sort_index().plot(kind='line', marker='o', ax=axes[1], color='green')
axes[1].set_title('Ocorrências por Ano')
axes[1].set_xlabel('Ano')
# Garante que o eixo X mostre apenas números inteiros para os anos
axes[1].set_xticks(df['Ano'].dropna().unique())

# Gráfico Linha - Hora (Agora agrupada e reduzida)
df_limpo['Hora_Agrupada'].value_counts().sort_index().plot(kind='line', marker='o', ax=axes[2], color='red')
axes[2].set_title('Ocorrências por Hora Cheia')
axes[2].set_xlabel('Hora')
axes[2].tick_params(axis='x', rotation=45) # Inclina os horários em 45 graus para caber bem

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("--- Estatísticas para: Idade da Vítima ---")

# 1. Força a conversão da coluna para números
df['Idade da Vítima Numérica'] = pd.to_numeric(df['Idade da Vítima'], errors='coerce')

# 2. Removemos os valores nulos para o cálculo
idade = df['Idade da Vítima Numérica'].dropna()

# 3. Calculando as métricas e organizando estruturadamente
estatisticas = {
    'Média (Posição)': round(idade.mean(), 2),
    'Mediana (Posição / Q2 / 50%)': round(idade.median(), 2),
    'Moda (Posição)': idade.mode()[0],
    'Desvio Padrão (Dispersão)': round(idade.std(), 2),
    'Variância (Dispersão)': round(idade.var(), 2),
    'Mínimo (Dispersão)': idade.min(),
    'Máximo (Dispersão)': idade.max(),
    'Amplitude (Dispersão)': idade.max() - idade.min(),
    '1º Quartil (Separatriz 25%)': round(idade.quantile(0.25), 2),
    '3º Quartil (Separatriz 75%)': round(idade.quantile(0.75), 2)
}

# 4. Criando a tabela (DataFrame) com os resultados
tabela_estatisticas = pd.DataFrame(list(estatisticas.items()), columns=['Medida Estatística', 'Valor'])

# 5. Exibindo a tabela formatada
display(tabela_estatisticas)

# 6. Plotando o Boxplot
plt.figure(figsize=(8, 4))
sns.boxplot(x=idade, color='#aec7e8')
plt.title('Boxplot - Distribuição da Idade da Vítima')
plt.xlabel('Idade')
plt.show()
```
